```python
import requests
from IPython.display import Image, display

titel = "Titanic"
url = f"http://www.omdbapi.com/?apikey={API_key}&t={titel}"

response = requests.get(url)
data = response.json()

if data.get("Response") == "False":
    print("Fout:", data.get("Error"))
else:
    # --- TEKST-dataset ---
    print("=== TEKST ===")
    print("Titel:", data["Title"])
    print("Jaar:", data["Year"])
    print("Plot:", data["Plot"])
    print("Genre:", data["Genre"])
    print("Cast:", data["Actors"])

    # --- BEELD-dataset ---
    print("\n=== BEELD ===")
    print("Poster URL:", data["Poster"])
    display(Image(url=data["Poster"]))
```

```python
import requests
from IPython.display import Audio, display

zoekterm = "My Heart Will Go On Celine Dion"
url = f"https://api.deezer.com/search?q={zoekterm}"

response = requests.get(url)
data = response.json()

if data.get("data"):
    for track in data["data"][:5]:
        print(track["title"], "-", track["artist"]["name"])

    eerste = data["data"][0]
    print("\nPreview van:", eerste["title"], "-", eerste["artist"]["name"])
    display(Audio(url=eerste["preview"]))
else:
    print("Geen resultaten gevonden")
```

```python
import requests
from bs4 import BeautifulSoup
import time

# --- Stap 1: inloggen via een sessie ---
session = requests.Session()

login_url = "https://example.com/login"
login_payload = {
    "username": "jouw_gebruikersnaam",
    "password": "jouw_wachtwoord"
}

login_response = session.post(login_url, data=login_payload)

if login_response.ok:
    print("Ingelogd, sessie/cookie actief.")
else:
    print("Login mislukt:", login_response.status_code)

# --- Stap 2: pagina ophalen mét de ingelogde sessie ---
time.sleep(1)  # respecteer de bron, geen requests-bombardement
target_url = "https://example.com/film/titanic/reviews"
pagina = session.get(target_url)

soup = BeautifulSoup(pagina.text, "html.parser")

# --- Stap 3: relevante tekst extraheren ---
reviews = []
for block in soup.select(".review-text"):  # pas CSS-selector aan op echte site
    tekst = block.get_text(strip=True)
    if tekst:
        reviews.append(tekst)

print(f"{len(reviews)} reviews gevonden")
print(reviews[:3])
```
